```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

fig = make_subplots(rows=2, cols=2, start_cell="bottom-left", shared_xaxes=True, 
                    subplot_titles=['Depression', 'Anxiety','Insomnia',  'OCD'])

filtered_df = data[(data['Depression'] > 7)]
# Count the occurrences of each level of improvement for the filtered data
improvement_counts = filtered_df['Music effects'].value_counts()

# Create a bar chart
fig.add_trace(go.Bar(x=improvement_counts.index,y=improvement_counts.values), 
              row=1, col=1)

filtered_df = data[(data['Anxiety'] > 7)]
# Count the occurrences of each level of improvement for the filtered data
improvement_counts = filtered_df['Music effects'].value_counts()
fig.add_trace(go.Bar( x=improvement_counts.index,y=improvement_counts.values),  
              row=1, col=2)

filtered_df = data[(data['Insomnia'] > 7)]
# Count the occurrences of each level of improvement for the filtered data
improvement_counts = filtered_df['Music effects'].value_counts()
fig.add_trace(go.Bar( x=improvement_counts.index,y=improvement_counts.values), 
              row=2, col=1)

filtered_df = data[(data['OCD'] > 7)]
# Count the occurrences of each level of improvement for the filtered data
improvement_counts = filtered_df['Music effects'].value_counts()
fig.add_trace(go.Bar( x=improvement_counts.index,y=improvement_counts.values),  
              row=2, col=2)
fig.update_traces(showlegend=False)
fig.update_layout(height=500, width=500,
                  title_text="Music Effects for different Mental Health Issues")
# Update y-axis range for both rows
for row in [1, 2]:
    for col in [1, 2]:
        fig.update_yaxes(range=[0, 200], row=row, col=col)
fig.show()
```

```python
bins = [10, 20, 30, 40, 50]
labels = ['10-20', '21-30', '31-40', '41-50']
data['age_group'] = pd.cut(data['Age'], bins=bins, labels=labels, right=False)

# Create a pivot table to get counts of each score for each age group
pivot_table = data.pivot_table(index='age_group', columns='Depression', aggfunc='size', fill_value=0, observed=False)

# Plotting the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(pivot_table, cmap='viridis', annot=True, fmt='d', cbar=True)
plt.xlabel('Depression Score')
plt.ylabel('Age Group')
plt.title('Count of People in Each Age Group Against Each Depression Score')
plt.show()
```

```python
# Create figure and subplots
fig = plt.figure(figsize=(10, 8))
plt.suptitle("Age vs Mental Health")

# Define data
y = data["Anxiety"]
y2 = data["Depression"]
y3 = data["Insomnia"]
y4 = data["OCD"]
x = data["Age"]

# Define number of bins
bins = 10

# Plot each histogram
for i, (label, y_data) in enumerate(zip(["Anxiety", "Depression", "Insomnia", "OCD"], [y, y2, y3, y4]), start=1):
    ax = fig.add_subplot(2, 2, i)
    plt.title(label)
    plt.xlabel('Age')
    plt.ylabel('Mental health ranking')
    heatmap, xedges, yedges = np.histogram2d(x, y_data, bins=bins)
    plt.hist2d(x, y_data, bins=bins, cmap='Blues')
    for i in range(len(xedges) - 1):
        for j in range(len(yedges) - 1):
            plt.text(xedges[i] + 0.5 * (xedges[i + 1] - xedges[i]), yedges[j] + 0.5 * (yedges[j + 1] - yedges[j]),
                     f'{heatmap[i, j]:.0f}', ha='center', va='center', color='white')

plt.tight_layout()
plt.show()


```
