```python
import random
guess = input("What is your guess, Heads or Tails?")
options = ("Heads", "Tails")
flip = random.choice(options)
if guess == flip:
    print("You have won!!")
else:
    print("Sorry, you did not guess correctly!")
```

```python
import ipywidgets as widgets
from IPython.display import display
import random

print("What is your guess, Heads or Tails?")
button_heads = widgets.Button(description="Heads")
button_tails = widgets.Button(description="Tails")

output = widgets.Output()

def game(b):
    options = ["Heads", "Tails"]
    flip = random.choice(options)

    with output:
        if b.description == flip:
            print("You have won!!")
        else:
            print("Sorry, you did not guess correctly!")


button_heads.on_click(game)
button_tails.on_click(game)

display(button_heads, button_tails, output)
```
